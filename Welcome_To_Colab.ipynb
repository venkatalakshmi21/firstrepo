```
import numpy as np
import matplotlib.pyplot as plt

# Define coordinates for the data points
x, y1, y2 = [1, 2, 3, 4, 5], [1, 4, 1, 3, 1], [1, 2, 1, 2, 1]

# Plot data tracks and define unique labels for the legend
plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')

# Customize the legend box appearance and location
plt.legend(title='Legend', loc='upper left', fontsize='small')

# Add descriptive titles and structural labels to the axes
plt.title('Customized Legend')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Render the visualization
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data since 'x' and 'y' aren't defined in the image snippet
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots()
line1, = ax.plot(x, y, label='sin(x)', color='tab:blue')
line2, = ax.plot(x, np.cos(x), label='cos(x)', color='tab:orange')

# 1. Create the first legend box and store it in a variable
legend1 = ax.legend(handles=[line1], loc='upper left')

# 2. Manually add the first legend back to the axes artist pool
ax.add_artist(legend1)

# 3. Create the second legend box at a different location
ax.legend(handles=[line2], loc='upper right')

plt.title('True Multiple Legends')
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

# 1. Initialize an empty figure window
plt.figure()

# 2. Create 1D arrays from -5 to 5 with 50 points, and combine them into a 2D coordinate grid
x_mesh, y_mesh = np.meshgrid(np.linspace(-5, 5, 50), np.linspace(-5, 5, 50))

# 3. Compute the Z values based on a sine function of the distance from the origin: z = sin(√(x² + y²))
z_mesh = np.sin(np.sqrt(x_mesh ** 2 + y_mesh ** 2))

# 4. Generate the 2D pseudocolor plot using the 'viridis' color map
plt.pcolormesh(x_mesh, y_mesh, z_mesh, cmap='viridis')

# 5. Display the color bar on the right showing the scalar-to-color mapping
plt.colorbar()

# 6. Add a title to the top of the plot
plt.title('Customized Color Bar')

# 7. Render and show the final plot on screen
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data to make the code fully executable
x = np.linspace(0, 10, 100)
y = np.sin(x) # Recreates the wave in Subplot 1

np.random.seed(42)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)

# Set up the overall figure canvas dimensions
plt.figure(figsize=(10, 8))

# Define and plot the first subplot (1 row, 2 columns, index 1)
plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title('Subplot 1')

# Define and plot the second subplot (1 row, 2 columns, index 2)
plt.subplot(2, 2, 2)
plt.scatter(x_scatter, y_scatter)
plt.title('Subplot 2')

# Display the final rendered subplots
plt.show()
```

## **Getting started**
</div>

The document you are reading is not a static web page, but an interactive environment called a **Colab notebook** that lets you write and execute code.

For example, here is a **code cell** with a short Python script that computes a value, stores it in a variable, and prints the result:

```
seconds_in_a_day = 24 * 60 * 60
seconds_in_a_day
```

To execute the code in the above cell, select it with a click and then either press the play button to the left of the code, or use the keyboard shortcut "Command/Ctrl+Enter". To edit the code, just click the cell and start editing.

Variables that you define in one cell can later be used in other cells:

```
seconds_in_a_week = 7 * seconds_in_a_day
seconds_in_a_week
```

Colab notebooks allow you to combine **executable code** and **rich text** in a single document, along with **images**, **HTML**, **LaTeX** and more. You can easily share your Colab notebooks with co-workers or friends, allowing them to comment on your notebooks or even edit them. ## Data science
</div>

With Colab you can harness the full power of popular Python libraries to analyze and visualize data. The code cell below uses **numpy** to generate some random data, and uses **matplotlib** to visualize it. To edit the code, just click the cell and start editing.

You can import your own data into Colab notebooks from your Google Drive account, including from spreadsheets, as well as from Github and many other sources. To learn more about importing data, and how Colab can be used for data science, see the links below under [Working with Data](#working-with-data).

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```
